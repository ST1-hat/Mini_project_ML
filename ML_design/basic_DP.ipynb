```python
import pandas as pd

import numpy as np
dt=pd.read_csv('required_files/apples_and_oranges.csv')

dt.head()
```

```python

dt.dtypes
dt['Class']=dt['Class'].map({'apple':0,'orange':1})
dt.head(50)
```

```python
from matplotlib import pyplot as plt
%matplotlib inline

df0=dt[dt.Class==0]
df1=dt[dt.Class==1]
plt.scatter(df0['Weight'],df0['Size'],color='red',marker='+')
plt.scatter(df1['Weight'],df1['Size'],color='orange',marker='*')
```

```python
from sklearn.model_selection import train_test_split
X=dt.drop(['Class'],axis='columns')
y=dt.Class
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
X_test.head(10)
```

```python
from sklearn.svm import SVC
model=SVC(C=10)

model.fit(X_train,y_train)
```

```python
train_score = model.score(X_train,y_train)
test_score = model.score(X_test,y_test)

print(f"Training Accuracy: {train_score}")
print(f"Test Accuracy: {test_score}")
```

```python
from sklearn.linear_model import LogisticRegression

log_model = LogisticRegression()
log_model.fit(X_train, y_train)
log_train=log_model.score(X_train, y_train)
log_test=log_model.score(X_test, y_test)
print(f"Logistic Regression Train Accuracy: {log_train}")   
print(f"Logistic Regression Test Accuracy: {log_test}")
```
